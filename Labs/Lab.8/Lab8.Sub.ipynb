### Exercise 3: Training a Classifier

Let's repeat what we did by hand in the previous lab (Lab 7, Exercise 8) using scikit-learn. We'll use an LDA classifier, which we can instantiate as follows:

```python
import sklearn.discriminant_analysis as DA
Fisher=DA.LinearDiscriminantAnalysis()
```

As discussed in the lecture, to properly formulate our problem, we'll have to:

* Define the inputs (X) vs outputs (Y)
* Designate training vs testing samples (in order to get an unbiased assessment of the performance of machine learning algorithms)

For example, here we'll use 4M events for training and the remainder for testing.

```python
N_Train=4000000

Train_Sample=df[:N_Train]
Test_Sample=df[N_Train:]

X_Train=Train_Sample[VarNames[1:]]
y_Train=Train_Sample["signal"]

X_Test=Test_Sample[VarNames[1:]]
y_Test=Test_Sample["signal"]

Test_sig=Test_Sample[Test_Sample.signal==1]
Test_bkg=Test_Sample[Test_Sample.signal==0]
```

We can train the classifier as follows:

```python
Fisher.fit(X_Train,y_Train)
```

We can plot the output, comparing signal and background:

```python
plt.figure()
plt.hist(Fisher.decision_function(Test_sig[VarNames[1:]]),bins=100,histtype="step", color="blue", label="signal",stacked=True)
plt.hist(Fisher.decision_function(Test_bkg[VarNames[1:]]),bins=100,histtype="step", color="red", label="background",stacked=True)
plt.legend(loc='upper right')
plt.show()
```

#### Part a

Compare the ROC curves computed on the test and on the training samples, in a single plot. Do you see a bias? (Use `Fisher.decision_function` on each sample, and `roc_curve` as in Exercise 5.)

```python
from sklearn.metrics import roc_curve, auc


y_train_scores = Fisher.decision_function(X_Train)
y_test_scores = Fisher.decision_function(X_Test)

fpr_train, tpr_train, _ = roc_curve(y_Train, y_train_scores)
fpr_test, tpr_test, _ = roc_curve(y_Test, y_test_scores)

plt.figure(figsize=(8, 6))
plt.plot(fpr_train, tpr_train, label=f"Train ROC (AUC = {auc(fpr_train, tpr_train):.4f})", color="blue", lw=2)
plt.plot(fpr_test, tpr_test, label=f"Test ROC (AUC = {auc(fpr_test, tpr_test):.4f})", color="red", linestyle="--", lw=2)
plt.plot([0, 1], [0, 1], "k--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Exercise 3a: Fisher Discriminant Train vs. Test ROC")
plt.legend(loc="lower right")
plt.grid(alpha=0.2)
plt.show()
```

```python
#There is virtually no bias (overfitting) between the training and testing ROC curves—they lie directly on top of one another. 
#Because LDA is a low-capacity linear model (only 18 weights and 1 bias parameter) trained on a massive sample size, variance is negligible.
```

#### Part b

Train three Fisher discriminants, using as input (1) only the raw quantities (`RawNames`), (2) only the derived features (`FeatureNames`), and (3) all 18. Compare their ROC curves, computed on the test sample, in a single plot. Which inputs carry the separation?

```python

plt.figure(figsize=(8, 6))

for label, cols in [("Raw Quantities (8)", RawNames), 
                    ("Derived Features (10)", FeatureNames), 
                    ("All Observables (18)", VarNames[1:])]:
    lda = DA.LinearDiscriminantAnalysis()
    lda.fit(Train_Sample[cols], y_Train)
    scores = lda.decision_function(Test_Sample[cols])
    fpr, tpr, _ = roc_curve(y_Test, scores)
    plt.plot(fpr, tpr, lw=2, label=f"{label} (AUC = {auc(fpr, tpr):.4f})")

plt.plot([0, 1], [0, 1], "k--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Exercise 3b: LDA ROC Comparison by Feature Set")
plt.legend(loc="lower right")
plt.grid(alpha=0.2)
plt.show()
```

```python
#The Derived Features (FeatureNames) carry most of the linear separation power (higher AUC than RawNames alone), 
#because high-level physics variables non-linearly combine raw 4-momenta into invariant/transverse mass quantities that a linear model cannot construct by itself from raw coordinates. 
#Combining all 18 inputs provides a slight further improvement.
```

#### Part c

Use the best method from part b to compute the maximal significance $\sigma_S= \frac{N_S}{\sqrt{N_S+N_B}}$ for the four scenarios in Lab 7 (Exercise 5.3), choosing the cut on the classifier's output that maximizes it.

```python

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.ensemble import HistGradientBoostingClassifier


N_fit = min(200000, len(Train_Sample))

classifiers = {
    "Logistic Regression": LogisticRegression(max_iter=500),
    "Gaussian Naive Bayes": GaussianNB(),
    "Gradient Boosting (HistGBDT)": HistGradientBoostingClassifier(max_iter=100, random_state=42)
}

def compare_classifier_inputs(clf, clf_name):
    plt.figure(figsize=(8, 6))
    best_scores = None
    best_auc = 0.0
    
    for label, cols in [("RawNames (8)", RawNames), 
                        ("FeatureNames (10)", FeatureNames), 
                        ("All 18", VarNames[1:])]:
        clf.fit(Train_Sample[cols][:N_fit], y_Train[:N_fit])
        
        if hasattr(clf, "decision_function"):
            scores = clf.decision_function(Test_Sample[cols])
        else:
            scores = clf.predict_proba(Test_Sample[cols])[:, 1]
            
        fpr, tpr, _ = roc_curve(y_Test, scores)
        roc_auc = auc(fpr, tpr)
        plt.plot(fpr, tpr, lw=2, label=f"{label} (AUC = {roc_auc:.4f})")
        
        if label == "All 18":
            best_scores = scores
            best_auc = roc_auc
            
    plt.plot([0, 1], [0, 1], "k--")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title(f"Exercise 4b: {clf_name} Input Comparison")
    plt.legend(loc="lower right")
    plt.grid(alpha=0.2)
    plt.show()
    return best_scores, best_auc

model_outputs = {}
for name, clf in classifiers.items():
    scores, roc_auc = compare_classifier_inputs(clf, name)
    model_outputs[name] = (scores, roc_auc)

best_model_name = max(model_outputs, key=lambda k: model_outputs[k][1])
best_scores, best_auc = model_outputs[best_model_name]
print(f"Best Classifier Selected: {best_model_name} (Test AUC = {best_auc:.4f})")

fpr_best, tpr_best, thresh_best = roc_curve(y_Test, best_scores)
scenarios = [(10, 100), (100, 1000), (1000, 10000), (10000, 100000)]

sig_rows = []
optimal_cuts_4c = {}

for Ns, Nb in scenarios:
    Ns_p = tpr_best * Ns
    Nb_p = fpr_best * Nb
    denom = np.sqrt(Ns_p + Nb_p)
    sigma = np.divide(Ns_p, denom, out=np.zeros_like(Ns_p), where=denom > 0)
    idx = np.argmax(sigma)
    
    optimal_cuts_4c[(Ns, Nb)] = (thresh_best[idx], tpr_best[idx], fpr_best[idx], sigma[idx])
    sig_rows.append([f"N_S={Ns}, N_B={Nb}", f"{thresh_best[idx]:.4f}", f"{tpr_best[idx]:.4f}", f"{fpr_best[idx]:.4f}", f"{sigma[idx]:.3f}"])

display(HTML(f"<h4>Exercise 4c: Maximal Significance ({best_model_name})</h4>"))
display(HTML(tabulate.tabulate(sig_rows, headers=["Scenario", "Optimal Cut", "TPR", "FPR", "Max Significance"], tablefmt="html")))
```

### Exercise 5: Metrics

Scikit-learn provides methods for computing the FPR, TPR, ROC, AUC metrics. For example:

```python
from sklearn.metrics import roc_curve, auc
fpr, tpr, _ = roc_curve(y_Test, Fisher.decision_function(X_Test))

roc_auc = auc(fpr, tpr)

plt.plot(fpr,tpr,color='darkorange',label='ROC curve (area = %0.2f)' % roc_auc)
plt.legend(loc="lower right")
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')

plt.show()
```

#### Part b
Look through [model evaluation](https://scikit-learn.org/stable/modules/model_evaluation.html#) documentation. Using scikit-learn's tools, compute TPR, FPR, ROC, AUC, Precision, Recall, F1 score, and accuracy for the method you selected in Exercise 4c and each scenario. Make a nice table, which also includes the maximal significance. 


```python

from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score

metrics_table = []

for (Ns, Nb), (cut, tpr_val, fpr_val, max_sig) in optimal_cuts_4c.items():
    y_pred = (best_scores >= cut).astype(int)
    
    prec = precision_score(y_Test, y_pred, zero_division=0)
    rec = recall_score(y_Test, y_pred, zero_division=0)
    f1 = f1_score(y_Test, y_pred, zero_division=0)
    acc = accuracy_score(y_Test, y_pred)
    
    metrics_table.append([
        f"N_S={Ns}, N_B={Nb}",
        f"{cut:.4f}",
        f"{tpr_val:.4f}",
        f"{fpr_val:.4f}",
        f"{best_auc:.4f}",
        f"{prec:.4f}",
        f"{rec:.4f}",
        f"{f1:.4f}",
        f"{acc:.4f}",
        f"{max_sig:.3f}"
    ])

headers = ["Scenario", "Cut", "TPR", "FPR", "AUC", "Precision", "Recall", "F1 Score", "Accuracy", "Max Significance"]
display(HTML(f"<h4>Exercise 5b: Evaluation Metrics Summary ({best_model_name})</h4>"))
display(HTML(tabulate.tabulate(metrics_table, headers=headers, tablefmt="html")))
```
