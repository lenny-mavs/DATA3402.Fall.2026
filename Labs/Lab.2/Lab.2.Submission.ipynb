*Exercise 4:* Write a function that takes an `n×n` matrix representing a tic‑tac‑toe game and returns one of the following values:

- `-1` if the game is **incomplete** (still empty spaces and no winner)
- `0` if the game is a **draw**
- `1` if **player 1** (`"X"`) has won
- `2` if **player 2** (`"O"`) has won

Here are some example inputs you can use to test your code:


````python
# Write your solution here
def game_status(board):
    n = len(board)
    
    for row in board:#First we check the rows
        if len(set(row)) == 1 and row[0] != 0:# By using set(), we elimninate the need for loops because it drops duplicates
            return row[0]
    for col_index in range(n):#Next we check the columns
        column = [board[row_index][col_index] for row_index in range(n)]
        if len(set(column)) == 1 and column[0] != 0: #similar logic pt.2 
            return column[0] 
    primary_diagonal = [board[i][i] for i in range(n)]#Now for the first diag
    if len(set(primary_diagonal)) == 1 and primary_diagonal[0] != 0:#similar logic pt.3
        return primary_diagonal[0]
    secondary_diagonal = [board[i][n - 1 - i] for i in range(n)]#Finally the second diag
    if len(set(secondary_diagonal)) == 1 and secondary_diagonal[0] != 0:#similar logic pt.4
        return secondary_diagonal[0]
    for row in board:
        if 0 in row:
            return -1 #No winners yet, the show must go on
    return 0
````

````python
winner_is_2 = [[2, 2, 0],
	[2, 1, 0],
	[2, 1, 1]]

winner_is_1 = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 1]]

winner_is_also_1 = [[0, 1, 0],
	[2, 1, 0],
	[2, 1, 1]]

no_winner = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 2]]

also_no_winner = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 0]]
````

````python
# Test your solution here
print(game_status(winner_is_2))
print(game_status(winner_is_1))
print(game_status(winner_is_also_1))
print(game_status(no_winner))
print(game_status(also_no_winner))
````

*Exercise 5:* Write a function that takes a game board, a player number, and `(row, col)` coordinates and places the correct mark (`"X"` or `"O"`) in that location.

Requirements:

- Only allow placing a mark in a previously empty location.
- Return `True` if the move was successful, and `False` otherwise.


````python
# Write your solution here
def place_mark(board, player, row, col):
    if board[row][col] == 0:
        board[row][col] = player
        return True
    else:
        return False
````

*Exercise 7:* Write a function that takes a board, a player number, and a location string (as in Exercise 6), then uses your function from Exercise 5 to update the board.

````python
# Write your solution here
def update_board_from_string(board, player, location):
    alphabet = "ABCDEFGHIJKLMNOPQRSTUVWXYZ"
    
    # Extract the letter and number
    row_letter = location[0].upper() # .upper() safely handles lowercase 
    col_string = location[1:]
    
    # Convert to indices
    row = alphabet.index(row_letter) #indexing to map letters to row values
    col = int(col_string) - 1
    #Send to board
    return place_mark(board, player, row, col)
````

*Exercise 8:* Write a function that is called with a board and player number, takes input from the player using Python's `input()`, and modifies the board using your function from Exercise 7.

Keep asking for input until the player enters a valid location that results in a valid move.


````python
# Write your solution here
def get_player_move(board, player):
    while True:
        # Ask the player for their initial input
        location = input(f"Player {player}, enter your move (e.g., A2): ")
        
        #Attempt placement
        success = update_board_from_string(board, player, location)
        
        # If successful, break the loop. Otherwise, let them try again.
        if success:
            print("Success!")
            break
        else:
            print("Invalid move or space taken. Please try again.")
````

*Exercise 9:* Use all of the previous exercises to implement a full tic‑tac‑toe game:

- draw the board,
- repeatedly ask two players for a location,
- apply valid moves,
- check the game status until a player wins or the game is a draw.


````python
# Write your solution here
def play_game(n):
    #Create a board and start player turn alternation
    board = create_board(n)
    is_player_one = True
    
    while True:
        #draw game board
        draw_game_board(board) 
        
        #current player based on your boolean
        current_player = 1 if is_player_one else 2
        
        # ask for a valid move
        get_player_move(board, current_player)
        
        #game status
        status = game_status(board)
        
        if status != -1:
            #gg
            draw_game_board(board)
            if status == 0:
                print("It's a draw!")
            else:
                print(f"Player {status} wins!")
            break # Exit the main game loop
            
        #Toggle the turn boolean
        is_player_one = not is_player_one
````

*Exercise 11:* Develop a version of the game where one player is the computer.

Note: you do **not** need an extensive search for the best move. For example, you can have the computer:
- block obvious losses
- otherwise try to create a winning row/column/diagonal


````python
# Write your solution here
def get_ai_move(board, computer_player):
    n = len(board)
    opponent = 1 if computer_player == 2 else 2

    #scanning logic similar to game programming
    def find_winning_move(target_player):
        # check rows
        for r in range(n):
            row = board[r]
            if row.count(target_player) == n - 1 and row.count(0) == 1:
                c = row.index(0) # The column is the location of the 0
                return r, c
                
        # check columns
        for c in range(n):
            col = [board[r][c] for r in range(n)]
            if col.count(target_player) == n - 1 and col.count(0) == 1:
                r = col.index(0) # The row is the location of the 0 in this extracted column
                return r, c
                
        # check diag 1
        diag1 = [board[i][i] for i in range(n)]
        if diag1.count(target_player) == n - 1 and diag1.count(0) == 1:
            i = diag1.index(0)
            return i, i # r and c are the same
            
        # check diag2
        diag2 = [board[i][n - 1 - i] for i in range(n)]
        if diag2.count(target_player) == n - 1 and diag2.count(0) == 1:
            i = diag2.index(0)
            return i, n - 1 - i # Math maps it back to the real coordinates
            
        return None # No winning move found 

    # is there a winning move?
    move = find_winning_move(computer_player)
    if move:
        return move

    # am i cooked?
    move = find_winning_move(opponent)
    if move:
        return move

    # whats left
    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                return r, c
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write your solution here
def scoring_ai(board, depth, current_player, computer_player):
    opponent = 1 if computer_player == 2 else 2

    status = game_status(board) 
    
    if status != -1: 
        if status == 0:
            return 0, None           
        elif status == computer_player:
            return 10, None          
        else:
            return -10, None         
            
    if depth == 0: 
        return 0, None 

    empty_spaces = []
    for r in range(len(board)):
        for c in range(len(board)):
            if board[r][c] == 0:
                empty_spaces.append((r, c))

    is_maximizing = (current_player == computer_player)
    
    if is_maximizing:
        best_score = float(-100)
    else:
        best_score = float(100)
        
    best_move = None

    for r, c in empty_spaces:
        #simulate a move
        board[r][c] = current_player
        
        #recursion 
        next_player = 1 if current_player == 2 else 2
        score, _ = scoring_ai(board, depth - 1, next_player, computer_player)
        
        #backtracing
        board[r][c] = 0
        
        #scoring
        if is_maximizing:
            if score > best_score:
                best_score = score
                best_move = (r, c)
        else:
            if score < best_score:
                best_score = score
                best_move = (r, c)

    return best_score, best_move
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
# Write your solution here
def simulate_games(n, num_games, depth_limit):
    scoring_ai_wins = 0
    ai_wins = 0
    draws = 0
    
    for game in range(num_games):
        #each game starts fresh board
        board = [[0 for _ in range(n)] for _ in range(n)] 
        current_player = 1  
        
        while True:
            if current_player == 1:
                # First AI's turn
                move = get_ai_move(board, current_player)
                r, c = move
                board[r][c] = current_player
            else:
                # Scoring AI's turn
                score, move = scoring_ai(board, depth_limit, current_player, 2)
                r, c = move
                board[r][c] = current_player
                
            # game_status
            status = game_status(board)
            if status != -1:
                if status == 2:
                    scoring_ai_wins += 1
                elif status == 1:
                    ai_wins += 1
                else:
                    draws += 1
                break # Exit this game's loop and start the next game
                
            # Toggle turn
            current_player = 1 if current_player == 2 else 2
            
    # win rate calculation
    win_rate = (scoring_ai_wins / num_games) * 100
    print(f"--- {n}x{n} Grid (Max Depth: {depth_limit}) ---")
    print(f"Scoring AI Win Rate: {win_rate}% ({scoring_ai_wins} wins, {draws} draws, {ai_wins} losses)")
````

````python
# Test your solution here
simulate_games(3,10,4)
simulate_games(4,10,4)
simulate_games(5,10,4)
````
